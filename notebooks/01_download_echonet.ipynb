```
from google.colab import userdata
import os

# NOTE: Redivis recommends NOT using an API token for interactive environments like Colab.
# Instead, the first call below will trigger an interactive OAuth login (a link + login prompt).
# If you previously set REDIVIS_API_TOKEN as a Colab secret, you can leave it — it will simply
# be ignored/unused now. Just make sure any token you created has been deleted on Redivis if
# it was ever pasted somewhere insecure.
print('Ready — the next cell will prompt an interactive Redivis login if needed.')
```

```
import redivis

# Using qualified reference IDs (from the earlier warning) avoids ambiguity if names change.
dataset = redivis.organization('AIMI').dataset('echonet_dynamic:66s1')
table = dataset.table('echonet:fjdn')
file_ref = table.file('EchoNet-Dynamic.zip')

zip_path = os.path.join(RAW_DIR, 'EchoNet-Dynamic.zip')
print('Downloading ~7GB — this can take a while on free Colab. Do not close the tab.')
print('If this prompts a login, follow the link and approve access, then re-run this cell.')
file_ref.download(zip_path)
print('Downloaded to', zip_path)
```

## Extract

7GB zipped will be larger unzipped. Extract directly onto Drive so it survives session disconnects — but note Drive has daily transfer quotas, so if this is slow/fails partway, resuming is safe (unzip skips existing files with `-n`).

```
extract_dir = os.path.join(RAW_DIR, 'EchoNet-Dynamic')
os.makedirs(extract_dir, exist_ok=True)
!unzip -n -q "{zip_path}" -d "{extract_dir}"
print('Extracted to', extract_dir)
print(os.listdir(extract_dir))
```

```
# Locate FileList.csv (metadata: EF, split, filenames) wherever it landed after extraction
import glob
candidates = glob.glob(os.path.join(extract_dir, '**', 'FileList.csv'), recursive=True)
print('Found FileList.csv at:', candidates)
```
